# performing logistic regression on penguins dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score,confusion_matrix, accuracy_score
from sklearn.preprocessing import LabelEncoder
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('penguins.csv')
df.head()
```

```python
df.drop(columns=['island','sex'],inplace=True)
df.isnull().sum()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df['body_mass_kg'] = df['body_mass_g']/1000
df.drop('body_mass_g',axis =1 , inplace =True)
df.head()
```

```python
df['bill_length_mm'] = df['bill_length_mm'].fillna(df['bill_length_mm'].mean())
df['bill_depth_mm'] = df['bill_depth_mm'].fillna(df['bill_depth_mm'].mean())
df['flipper_length_mm'] = df['flipper_length_mm'].fillna(df['flipper_length_mm'].mean())
df['body_mass_kg'] = df['body_mass_kg'].fillna(df['body_mass_kg'].mean())
```

```python
df.isnull().sum()
```

# Now the data is cleaned

```python
x = df.drop('species',axis=1)
y =df['species']
```

```python
le = LabelEncoder()
y = le.fit_transform(y)
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=24)
```

```python
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_test
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
print('accuracy score:', accuracy_score(y_test,y_pred))
```

```python
print('confusion matrix:\n', confusion_matrix(y_test,y_pred))
```

```python
print(mean_absolute_error(y_test,y_pred))
```

```python
print(mean_squared_error(y_test,y_pred))
```

```python
print(r2_score(y_test,y_pred))
```
